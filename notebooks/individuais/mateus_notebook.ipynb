```python
import pandas as pd
import numpy as np

caminho = r"D:\mateu\Documents\machine-learning-project\data\Final_Augmented_dataset_Diseases_and_Symptoms.csv"

print("Lendo o arquivo gigante do disco...")
df = pd.read_csv(caminho)
print(f"Tamanho total lido: {df.shape}")

df_micro = df.sample(n=1000, random_state=42)
print(f"Tamanho da amostra separada: {df_micro.shape}")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score

X = df_micro.drop(columns=['diseases'])
y = df_micro['diseases']

# A Árvore de Decisão clássica (super leve)
model = DecisionTreeClassifier(random_state=42)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("Iniciando o treino relâmpago...")
results = cross_val_score(model, X, y, cv=skf, scoring='accuracy')

print(f"Resultados: {results}")
print(f"Acurácia Média: {np.mean(results) * 100:.2f}%")
```
